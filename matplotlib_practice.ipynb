```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

NAME = 'leena bachhav'
ROLL_NO = 20
```

```python
assert NAME != '' and ROLL_NO != 1, "You forgot to enter your NAME and ROLL_NO!"

np.random.seed(ROLL_NO)
days = np.arange(1, 16)
study = np.random.randint(1, 8, 15)
marks = study*10 + np.random.randint(0, 11, 15) + (ROLL_NO % 5)
sleep = np.random.randint(5, 9, 15)
screen = np.random.randint(2, 8, 15)
water = np.random.randint(2, 6, 15)

df = pd.DataFrame({
    'Day': days, 'Study Hours': study, 'Marks': marks,
    'Sleep Hours': sleep, 'Screen Time': screen, 'Water Intake': water,
    'Assignment_ID': ROLL_NO
})

print(f"Dataset generated for: {NAME}  |  Roll No: {ROLL_NO}  |  Assignment_ID: {ROLL_NO}")
df
```

```python
plt.plot(df["Day"],df["Marks"],marker="o")
plt.title("day vs marks")
plt.xlabel("day")
plt.ylabel("marks")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.bar(df["Day"],df["Study Hours"],color="red",edgecolor="black")
plt.title("day vs study")
plt.show()

average_study_hours=df["Study Hours"].mean()
average_study_hours
```

#scatter plot

```python
plt.scatter(df["Study Hours"],df["Marks"],color="indigo")
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid()
plt.show()

correlation=df["Study Hours"].corr(df["Marks"])
correlation
```

```python
plt.hist(df["Marks"],bins=7,color="green",edgecolor="black")
plt.title("histogram")
plt.xlabel("x")
plt.ylabel("Y")
plt.show()
```

```python
import matplotlib.pyplot as plt

avg_values = [
    df["Study Hours"].mean(),
    df["Sleep Hours"].mean(),
    df["Screen Time"].mean(),
    df["Water Intake"].mean()
]

labels = ["Study Hours", "Sleep Hours", "Screen Time", "Water Intake"]

plt.pie(avg_values, labels=labels, autopct="%1.1f%%", startangle=90)
plt.title("Average Daily Activities")
plt.show()

print("Average Study Hours:", df["Study Hours"].mean())
print("Average Sleep Hours:", df["Sleep Hours"].mean())
print("Average Screen Time:", df["Screen Time"].mean())
print("Average Water Intake:", df["Water Intake"].mean())
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 9))
axes[0, 0].plot(df["Day"], df["Marks"], marker="o")
axes[0, 0].set_title("Day vs Marks")
axes[0, 0].set_xlabel("Day")
axes[0, 0].set_ylabel("Marks")
axes[0, 0].grid()
axes[0, 1].bar(df["Day"], df["Study Hours"], edgecolor="black")
axes[0, 1].set_title("Day vs Study Hours")
axes[0, 1].set_xlabel("Day")
axes[0, 1].set_ylabel("Study Hours")

# 3. Study Hours vs Marks
axes[1, 0].scatter(df["Study Hours"], df["Marks"])
axes[1, 0].set_title("Study Hours vs Marks")
axes[1, 0].set_xlabel("Study Hours")
axes[1, 0].set_ylabel("Marks")
axes[1, 0].grid()

# 4. Average Activity Values
avg_values = [
    df["Study Hours"].mean(),
    df["Sleep Hours"].mean(),
    df["Screen Time"].mean(),
    df["Water Intake"].mean()
]

labels = ["Study", "Sleep", "Screen", "Water"]

axes[1, 1].pie(
    avg_values,
    labels=labels,
    autopct="%1.1f%%"
)
axes[1, 1].set_title("Average Daily Activities")

# Main dashboard title
fig.suptitle("Student Performance Dashboard", fontsize=18)

plt.tight_layout()
plt.show()
```
